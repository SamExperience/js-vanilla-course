### LESSON 24 — Exercise

```typescript
const quantity = "3";
const unitPrice = "19.99";

console.log(quantity + unitPrice);            // "319.99" — both are text

const total = Number(quantity) + Number(unitPrice);
console.log(total);
console.log(total.toFixed(2));

console.log(parseInt("12kg"));
console.log(Number.isNaN(Number("abc")));
```

### LESSON 24 — Mini challenge

```typescript
function rollDice() {
  return Math.floor(Math.random() * 6) + 1;
}

console.log(rollDice(), rollDice(), rollDice());

const temperatures = [18.4, 21.9, 15.2];

let highest = temperatures[0];
let lowest = temperatures[0];

for (const value of temperatures) {
  if (value > highest) highest = value;
  if (value < lowest) lowest = value;
}

console.log(highest, lowest);

function toCents(euros) {
  return Math.round(euros * 100);
}

console.log(toCents(19.99));
```
