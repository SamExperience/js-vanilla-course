# JavaScript — Numbers and Math

## LESSON 24 — Numbers and Math

Text that looks like a number is still text. `"5" + 1` is `"51"`, because `+` on a string means "join". Converting is your job.

| from text to number | what it does |
|---|---|
| `Number(text)` | converts the whole string, or gives `NaN` |
| `parseInt(text)` | reads an integer from the front and stops |
| `parseFloat(text)` | same, but keeps the decimals |

```js
Number("42");        // 42
Number("42px");      // NaN     -> the whole string has to be a number
parseInt("42px");    // 42      -> reads until it cannot
parseFloat("3.5kg"); // 3.5
```

An `<input>` always hands you a string, even `type="number"`. This is where that matters.

### NaN

`NaN` means "not a number", and it is what you get when a calculation makes no sense.

```js
Number("hello");        // NaN
NaN === NaN;            // false   <- yes, really
Number.isNaN(value);    // the only reliable way to test for it
```

### Rounding and formatting

```js
Math.round(4.5);      // 5     -> nearest
Math.floor(4.9);      // 4     -> always down
Math.ceil(4.1);       // 5     -> always up
(3.14159).toFixed(2); // "3.14"   <- a STRING, not a number
```

### The Math object

| call | gives back |
|---|---|
| `Math.min(a, b)` / `Math.max(a, b)` | the smaller / larger one |
| `Math.abs(n)` | the value without its sign |
| `Math.random()` | a decimal from 0 up to (but never) 1 |
| `Math.pow(a, b)` | `a` to the power of `b` — or write `a ** b` |

A random whole number from 1 to 6:

```js
Math.floor(Math.random() * 6) + 1;
```

### Key notes

- **`toFixed` returns a string.** `(2.5).toFixed(2) + 1` gives `"2.501"`. Wrap it in `Number()` if you need to keep calculating.
- **Never compare with `NaN` using `===`.** `NaN === NaN` is `false`. Use `Number.isNaN(x)`.
- `0.1 + 0.2` is `0.30000000000000004`. That is not a bug in JavaScript, it is how computers store decimals. For money, work in whole cents, or round at the moment you display.
- `Math.random()` never returns 1, which is why the dice formula uses `* 6` and not `* 7`.

### Example

````typescript
const exampleInput = "25";

console.log(exampleInput + 1);
console.log(Number(exampleInput) + 1);

console.log(parseInt("42px"));
console.log(parseFloat("3.5kg"));
console.log(Number("hello"));
console.log(Number.isNaN(Number("hello")));

console.log(Math.round(4.5), Math.floor(4.9), Math.ceil(4.1));
console.log((3.14159).toFixed(2));
console.log(Math.min(3, 9), Math.max(3, 9), Math.abs(-7));
console.log(0.1 + 0.2);
````
