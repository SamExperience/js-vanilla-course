# JavaScript — Map, Set and friends

## LESSON 35 — Map and Set

Objects and arrays cover most needs. Two other collections cover the cases where they are awkward.

### Set — a list with no duplicates

```js
const tags = new Set(["js", "css", "js"]);

tags.size;            // 2      <- the duplicate never got in
tags.has("css");      // true
tags.add("html");
tags.delete("css");
[...tags];            // ["js", "html"]   <- back to an array
```

Removing duplicates from an array is one line:

```js
[...new Set([1, 2, 2, 3])];    // [1, 2, 3]
```

### Map — an object whose keys can be anything

An object's keys are always strings. A `Map` keeps whatever you give it, and remembers the order you inserted things in.

```js
const prices = new Map();

prices.set("laptop", 1200);
prices.set("mouse", 25);

prices.get("laptop");     // 1200
prices.has("mouse");      // true
prices.size;              // 2
prices.delete("mouse");

for (const [name, price] of prices) {
  console.log(name, price);
}
```

| you want | object | Map |
|---|---|---|
| string keys, fixed shape | yes | overkill |
| keys added and removed constantly | awkward | yes |
| non-string keys | impossible | yes |
| how many entries? | `Object.keys(o).length` | `.size` |

### Key notes

- **`Set` compares by identity, not by content.** Two objects that look identical are two different values, so a `Set` keeps both.
- `map.get(key)` returns `undefined` when the key is missing — the same as an object.
- A `Map` is **iterable in insertion order**; a plain object gives no such promise for every kind of key.
- Both convert back to an array with spread: `[...set]`, `[...map]` — and `[...map]` gives you `[key, value]` pairs.

### Example

````typescript
const exampleTags = new Set(["js", "css", "js", "html"]);

console.log(exampleTags.size);
console.log(exampleTags.has("css"));
console.log([...exampleTags]);

const exampleNumbers = [1, 2, 2, 3, 3, 3];
console.log([...new Set(exampleNumbers)]);

const exampleStock = new Map();
exampleStock.set("laptop", 4);
exampleStock.set("mouse", 12);

console.log(exampleStock.get("laptop"));
console.log(exampleStock.size);

for (const [item, quantity] of exampleStock) {
  console.log(`${item}: ${quantity}`);
}
````

## LESSON 36 — WeakMap, WeakSet, Symbol and BigInt

Four things you will read in other people's code long before you write them yourself. Know what they are; do not go looking for reasons to use them.

### WeakMap and WeakSet

They work like `Map` and `Set`, with three differences: **keys must be objects**, you cannot loop over them, and they do not keep their keys alive. When nothing else in your program refers to a key object, the entry disappears on its own.

```js
const metadata = new WeakMap();
let element = { id: 1 };

metadata.set(element, { clicks: 0 });
metadata.get(element);        // { clicks: 0 }

element = null;               // the entry is now free to be cleaned up
```

That is the whole point: attaching data to an object without preventing it from being freed. Libraries use this; application code rarely needs to.

### Symbol — a value guaranteed to be unique

```js
const id = Symbol("id");
const other = Symbol("id");

id === other;      // false   <- the text is only a label, for debugging
```

Used as an object key, a symbol cannot collide with anything else and does not show up in `Object.keys` or `JSON.stringify`. That is how a library adds a key to your object without any chance of clashing with yours.

### BigInt — integers beyond the safe range

Regular numbers stop being exact past `Number.MAX_SAFE_INTEGER` (about 9 quadrillion).

```js
Number.MAX_SAFE_INTEGER;        // 9007199254740991
9007199254740993;               // 9007199254740992   <- silently wrong

const big = 9007199254740993n;  // the n suffix makes it a BigInt
big + 1n;                       // exact
```

### Key notes

- **A `WeakMap` cannot be inspected.** No `.size`, no looping, no listing keys. If you want to see what is in it, you wanted a `Map`.
- A `Symbol` used as a key is **skipped** by `Object.keys`, `for...in` and `JSON.stringify`. That is a feature, and also a good way to lose data by accident.
- **BigInt and Number do not mix.** `1n + 1` throws. Convert one side deliberately.
- If you cannot name a concrete reason to use one of these four, the plain version is the right choice.

### Example

````typescript
const exampleCache = new WeakMap();
let exampleUser = { name: "Mia" };

exampleCache.set(exampleUser, { lastSeen: "today" });
console.log(exampleCache.get(exampleUser));
console.log(exampleCache.has(exampleUser));

const exampleId = Symbol("id");
const exampleOther = Symbol("id");
console.log(exampleId === exampleOther);

const exampleRecord = { name: "Mia", [exampleId]: 123 };
console.log(Object.keys(exampleRecord));
console.log(JSON.stringify(exampleRecord));
console.log(exampleRecord[exampleId]);

console.log(Number.MAX_SAFE_INTEGER);
console.log(9007199254740993);
console.log(9007199254740993n + 1n);
````
