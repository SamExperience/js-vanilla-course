# Solutions — Map, Set and friends

Only open this after you have actually tried. Run the cells in order: some reuse a variable from an earlier one.

### LESSON 35 — Exercise

```typescript
const visits = ["mia", "alex", "mia", "sam", "alex", "mia"];

const unique = new Set(visits);
console.log(unique.size);
console.log([...unique]);

const counts = new Map();

for (const name of visits) {
  counts.set(name, (counts.get(name) ?? 0) + 1);
}

console.log(counts.get("mia"));

for (const [name, count] of counts) {
  console.log(`${name}: ${count}`);
}
```

### LESSON 35 — Mini challenge

```typescript
const a = [1, 2, 3, 4, 5];
const b = [4, 5, 6, 7];

const setB = new Set(b);

console.log(a.filter((value) => setB.has(value)));
console.log(a.filter((value) => !setB.has(value)));
console.log([...new Set([...a, ...b])]);
```

### LESSON 36 — Exercise

```typescript
const secret = Symbol("secret");

const record = { name: "Mia", [secret]: "hidden value" };

console.log(Object.keys(record));      // only "name" — the symbol is skipped
console.log(record[secret]);

console.log(Number.MAX_SAFE_INTEGER);
console.log(Number.MAX_SAFE_INTEGER + 2);          // not exact
console.log(BigInt(Number.MAX_SAFE_INTEGER) + 2n); // exact
```

### LESSON 36 — Mini challenge

```typescript
const clicks = new WeakMap();

function registerClick(target) {
  clicks.set(target, (clicks.get(target) ?? 0) + 1);
}

const buttonA = { id: "a" };
const buttonB = { id: "b" };

registerClick(buttonA);
registerClick(buttonA);
registerClick(buttonA);
registerClick(buttonB);

console.log(clicks.get(buttonA));
console.log(clicks.get(buttonB));

// A Map would keep buttonA and buttonB alive for as long as the map exists,
// even after the rest of the program had finished with them.
```
