### LESSON 29 — Exercise

```typescript
const prices = [10, 25, 4, 60, 33];

console.log(prices.map((price) => price * 2));
console.log(prices.filter((price) => price > 20));
console.log(prices.find((price) => price > 20));
console.log(prices.reduce((total, price) => total + price, 0));
```

### LESSON 29 — Mini challenge

```typescript
const users = [
  { name: "Sam", age: 34, active: true },
  { name: "Alex", age: 17, active: false },
  { name: "Mia", age: 28, active: true },
];

console.log(users.map((user) => user.name));
console.log(users.filter((user) => user.active));
console.log(users.find((user) => user.name === "Mia"));
console.log(users.reduce((total, user) => total + user.age, 0));
```

### LESSON 30 — Exercise

```typescript
const orders = [
  { id: 1, total: 40, paid: true },
  { id: 2, total: 15, paid: false },
  { id: 3, total: 90, paid: true },
];

console.log(orders.some((order) => !order.paid));
console.log(orders.every((order) => order.total > 10));
console.log(orders.findIndex((order) => order.total > 50));

console.log(orders.map((order) => `${order.id}: ${order.total}`).join(", "));

console.log(
  orders
    .filter((order) => order.paid)
    .reduce((sum, order) => sum + order.total, 0),
);
```

### LESSON 30 — Mini challenge

```typescript
const classes = [
  { name: "A", students: ["mia", "alex"] },
  { name: "B", students: ["sam"] },
  { name: "C", students: [] },
];

console.log(classes.flatMap((group) => group.students));

console.log(
  classes.filter((group) => group.students.length > 0).map((group) => group.name),
);

console.log(classes.every((group) => group.students.length < 5));

console.log(
  classes.map((group) => `${group.name} (${group.students.length})`).join(" | "),
);
```
