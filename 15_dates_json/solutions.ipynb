### LESSON 37 — Mini challenge

```typescript
function daysUntil(target) {
  const oneDay = 1000 * 60 * 60 * 24;
  const today = new Date();
  return Math.round((target.getTime() - today.getTime()) / oneDay);
}

console.log(daysUntil(new Date(2030, 0, 1)));

function isWeekend(date) {
  const day = date.getDay();
  return day === 0 || day === 6;
}

console.log(isWeekend(new Date(2026, 8, 12)));   // Saturday
console.log(isWeekend(new Date(2026, 8, 14)));   // Monday
```

### LESSON 38 — Exercise

```typescript
const order = { id: 7, items: ["pen", "book"], total: 24.5 };

const text = JSON.stringify(order);
console.log(text);
console.log(JSON.stringify(order, null, 2));

const back = JSON.parse(text);
console.log(back.items[0]);

console.log(typeof text, typeof back);

console.log(JSON.stringify({ save: () => {}, note: undefined, id: 1 }));
// Only id survived: functions and undefined are dropped without warning.
```

### LESSON 38 — Mini challenge

```typescript
function safeParse(text) {
  try {
    return JSON.parse(text);
  } catch {
    return null;
  }
}

console.log(safeParse('{"ok":true}'));
console.log(safeParse("oops"));

const withDate = JSON.stringify({ when: new Date(2026, 0, 1) });
const parsed = JSON.parse(withDate);

console.log(typeof parsed.when);        // "string" — the Date did not survive
console.log(new Date(parsed.when).getFullYear());
```
