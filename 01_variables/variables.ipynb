# JavaScript — Variables

Types covered here: `string`, `number`, `boolean`, `null`, `undefined`, plus `typeof`.

## LESSON 1 — let & const

- `let` → variable, its value **can** be reassigned.
- `const` → constant, its value **cannot** be reassigned.

### Example

````typescript
let age = 33;
age = 34;

const name = "Sam";

console.log(age, name);
````

## LESSON 4 — Booleans

A **boolean** is either `true` or `false`.

- `true` → yes / active
- `false` → no / inactive

**Key note:** `true` and `false` are NOT strings — no quotation marks.

### Example

````typescript
const exampleLoggedIn = true;
const examplePermission = false;

console.log(exampleLoggedIn);
console.log(examplePermission);
````

## LESSON 6 — typeof

`typeof` tells you the type of a value.

```js
typeof age;   // -> "number"
```

### Exercise

4. Print the type of each of the three variables.

````typescript
// Your code here
let myAge = "34";
let myName = "Mario";
let isStudent = false;
console.log(typeof(myAge),typeof(myName),typeof(isStudent));
````
