## LESSON 14 — for loop

A `for` loop repeats code a specific number of times.

```js
for (initialization; condition; increment) {
    // code
}
```

```js
for (let i = 0; i < 5; i++) {
    console.log(i);
}
```

**How it runs:**

1. `let i = 0` → initialization
2. `i < 5` → condition
3. execute the code
4. `i++` → increment
5. repeat from step 2

### Key notes

`i++` means `i = i + 1`.

`<` determines when the loop stops. `i < 5` produces `0 1 2 3 4` — the value **5 is NOT included**.

### Example

````typescript
for (let exampleCounter = 0; exampleCounter < 5; exampleCounter++) {
  console.log(exampleCounter);
}
````

## LESSON 15 — while loop

A `while` loop repeats code as long as a condition is true.

```js
while (condition) {
    // code
}
```

### Key notes

`while` checks the condition **before** each iteration.

The variable must change inside the loop, otherwise the loop becomes infinite.

```js
let counter = 0;

while (counter < 5) {
  console.log(counter);
  counter++;      // without this line: infinite loop
}
```

- `for` → convenient when you know how many iterations you need.
- `while` → useful when the loop continues until a condition becomes false.

### Example

````typescript
let exampleCounterWhile = 0;

while (exampleCounterWhile < 5) {
  console.log(exampleCounterWhile);
  exampleCounterWhile++;
}
````

## LESSON 16 — break & continue

- `break` → immediately stops the loop.
- `continue` → skips the current iteration and continues with the next one.

### Key notes

Think of it as:

- `break` = **STOP the loop**
- `continue` = **SKIP this iteration**

### Example

````typescript
// Example — break. Output: 1 2 3 4 5
for (let exampleBreakNumber = 1; exampleBreakNumber <= 10; exampleBreakNumber++) {
  if (exampleBreakNumber === 6) {
    break;
  }

  console.log(exampleBreakNumber);
}
````

````typescript
// Example — continue. Output: 1 2 4 5
for (let exampleContinueNumber = 1; exampleContinueNumber <= 5; exampleContinueNumber++) {
  if (exampleContinueNumber === 3) {
    continue;
  }

  console.log(exampleContinueNumber);
}
````

## LESSON 17 — for...of

`for...of` iterates over the **values** of an iterable, such as an array or a string.

```js
for (const value of collection) {
    // code
}
```

### Key notes

With an array, `for...of` gives you each VALUE:

```js
const numbers = [10, 20, 30];

for (const number of numbers) {
  console.log(number);   // 10, 20, 30
}
```

This differs from a classic `for` loop, where you usually work with an index:

```js
for (let i = 0; i < numbers.length; i++) {
  console.log(numbers[i]);
}
```

### Example

````typescript
const exampleFruits = ["Apple", "Banana", "Orange"];

for (const exampleFruit of exampleFruits) {
  console.log(exampleFruit);
}
````

## LESSON 18 — for...in

`for...in` iterates over the **keys** of an object. With arrays, it gives you the **indexes**.

```js
for (const key in object) {
    // code
}
```

To access the value, use `object[key]`.

### for...in vs for...of

| | `for...in` | `for...of` |
|---|---|---|
| Object | keys | values (only if iterable) |
| Array | indexes | values |

**Key rule:** `for...in` → keys / indexes &nbsp;·&nbsp; `for...of` → values

### Example

````typescript
const exampleUser = {
  name: "Sam",
  age: 33,
  role: "Developer",
};

// Just the keys
for (const exampleKey in exampleUser) {
  console.log(exampleKey);
}

// Keys and values
for (const exampleUserKey in exampleUser) {
  console.log(exampleUserKey, exampleUser[exampleUserKey]);
}
````
