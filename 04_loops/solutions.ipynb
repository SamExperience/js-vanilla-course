# Solutions — Loops

Only open this after you have actually tried. Run the cells in order: some reuse a variable from an earlier one.

### LESSON 14 — Exercise

```typescript
for (let numbers = 1; numbers < 6; numbers++) {
  console.log(numbers);
}
```

### LESSON 14 — Mini challenge

```typescript
for (let numbers = 0; numbers <= 10; numbers++) {
  if (numbers % 2 === 0) console.log(numbers);
}
```

### LESSON 15 — Exercise

```typescript
let countdown = 5;

while (countdown > 0) {
  console.log(countdown);
  countdown--;
}
```

### LESSON 15 — Mini challenge

```typescript
let number = 1;

while (number <= 10) {
  console.log(number);
  number++;
}
```

### LESSON 16 — Exercise

```typescript
for (let numbers = 1; numbers <= 10; numbers++) {
  if (numbers === 7) break;
  console.log(numbers);
}
```

### LESSON 16 — Mini challenge

```typescript
for (let numbers = 1; numbers <= 10; numbers++) {
  if (numbers === 5) continue;
  console.log(numbers);
}
```

### LESSON 17 — Exercise

```typescript
const loopFruits = ["Apple", "Banana", "Orange"];

for (const fruit of loopFruits) {
  console.log(fruit);
}
```
