### LESSON 40 — what `lib/string-utils.js` should contain

```js
export function capitalize(word) {
  return word[0].toUpperCase() + word.slice(1);
}

export function reverse(word) {
  return word.split("").reverse().join("");
}

export default function slugify(text) {
  return text.toLowerCase().replaceAll(" ", "-");
}
```

### LESSON 40 — Exercise

Named exports: imported **with** braces, names must match exactly.

````typescript
import { capitalize, reverse } from "./solutions/string-utils.js";

console.log(capitalize("javascript"));
console.log(reverse("vite"));
````
