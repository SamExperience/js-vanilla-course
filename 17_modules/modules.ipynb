# JavaScript — Modules

## LESSON 40 — export & import

Until now everything lived in one place. A **module** is just a file that decides what to share with other files. This is the idea the whole Vite and React folder structure is built on.

### Named exports

```js
// lib/math-utils.js
export function add(a, b) {
  return a + b;
}
```

```js
import { add } from "./lib/math-utils.js";
```

The braces are not decoration: you're picking specific names out of the file.

### Default export

A file can have **one** default export. Import it without braces, under any name you like.

```js
export default function describe() { ... }
```

```js
import describe from "./lib/math-utils.js";        // no braces
import whateverIWant from "./lib/math-utils.js";   // also fine
```

### Mixing both, and renaming

```js
import describe, { add, multiply } from "./lib/math-utils.js";
import { add as sum } from "./lib/math-utils.js";
```

### Why bother

Anything **not** exported stays private to its file. That's the whole point: each file owns one job, and shows only what others need. In the Vite projects coming next you'll see this as `src/logic/`, `src/ui/`, `src/services/`.

### Key notes

- **Named import → braces. Default import → no braces.** Getting this backwards is the number one module error.
- A named import must match the exported name **exactly** (unless you use `as`). A default import can be called anything.
- In the browser and in Deno the path needs the **`.js` extension** and must start with `./` or `../`.
- Not exported = not importable. It isn't hidden by accident, it's private by design.

### Example

Open `lib/math-utils.js` and `lib/greetings.js` next to this notebook first — the code below imports from them for real.

````typescript
import describe, { add, multiply } from "./lib/math-utils.js";
import { greet, shout, GREETING } from "./lib/greetings.js";

console.log(add(2, 3));
console.log(multiply(4, 5));

// default export: imported without braces, name is ours to choose
console.log(describe());

console.log(greet("Sam"));
console.log(shout("Sam"));
console.log(GREETING);
````
