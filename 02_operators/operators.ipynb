# JavaScript — Operators

## LESSON 7 — Cheat sheet

**Arithmetic**

| | |
|---|---|
| `+` | addition |
| `-` | subtraction |
| `*` | multiplication |
| `/` | division |
| `%` | remainder |

**Comparison**

| | |
|---|---|
| `>` | greater than |
| `<` | less than |
| `>=` | greater or equal |
| `<=` | less or equal |
| `===` | strict equality |
| `!==` | strict difference |

**Logical**

| | |
|---|---|
| `&&` | AND |
| `\|\|` | OR |
| `!` | NOT |

**Assignment**

| | |
|---|---|
| `=` | assign |
| `+=` | add and assign |
| `-=` | subtract and assign |

## LESSON 8 — Key notes

**Prefer `===` over `==`**

```js
5 === "5"   // false
5 == "5"    // true
```

**`%` returns the REMAINDER**

```js
10 / 3   // 3.33...
10 % 3   // 1
```

**Important:** JavaScript does NOT support chained comparisons.

```js
40 >= score <= 50           // wrong
score >= 40 && score <= 50  // correct
```

## LESSON 9 — Operators in action

### Example

````typescript
const exampleA = 20;
const exampleB = 6;

console.log(exampleA + exampleB);
console.log(exampleA - exampleB);
console.log(exampleA * exampleB);
console.log(exampleA / exampleB);
console.log(exampleA % exampleB);

console.log(exampleA > exampleB);
console.log(exampleA === exampleB);
console.log(exampleA !== exampleB);

console.log(true && false);
console.log(true || false);
console.log(!true);
````
