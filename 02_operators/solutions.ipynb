# Solutions — Operators

Only open this after you have actually tried. Run the cells in order: some reuse a variable from an earlier one.

### LESSON 9 — Exercise

```typescript
let userAge = 25;
let minimumAge = 18;
let wallet = 120;
let itemPrice = 35;

console.log(wallet + itemPrice);
console.log(wallet - itemPrice);
console.log(wallet * 2);
console.log(wallet / itemPrice);
console.log(wallet % itemPrice);

// Is userAge greater than minimumAge?
console.log(userAge > minimumAge);
// Is userAge equal to 25? (strict)
console.log(userAge === 25);
// Is wallet different from 100? (strict)
console.log(wallet !== 100);
// Can buy?
console.log(wallet >= itemPrice && userAge >= minimumAge);
```

### LESSON 9 — Mini challenge

```typescript
let score = 47;

// 1. Is score even?
console.log(score % 2 === 0);

// 2. Is score greater than 40?
console.log(score > 40);

// 3. Is score between 40 and 50?
console.log(score >= 40 && score <= 50);
```
