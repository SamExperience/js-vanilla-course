### LESSON 27 — Exercise

```typescript
const book = {
  title: "1984",
  author: "Orwell",
  year: 1949,
};

console.log(book.title);

book.available = true;
book.year = 1950;
delete book.author;

console.log(Object.keys(book));

const sameBook = { title: "1984", year: 1950, available: true };

console.log(sameBook === book);   // false: same content, but a different object
```

### LESSON 27 — Mini challenge

```typescript
const cart = {
  items: ["pen", "notebook"],

  addItem(name) {
    this.items.push(name);
  },

  count() {
    return this.items.length;
  },
};

cart.addItem("eraser");

console.log(cart.count());
console.log(cart.count);   // the function itself: without () nothing is called
```

### LESSON 28 — Exercise

```typescript
const settings = { theme: "light", fontSize: 14, notifications: true };

console.log(Object.keys(settings));
console.log(Object.values(settings));
console.log(Object.entries(settings));

for (const [key, value] of Object.entries(settings)) {
  console.log(`${key} = ${value}`);
}

console.log({ ...settings, theme: "dark" });
console.log(settings);

console.log(Object.hasOwn(settings, "language"));

const field = "language";
console.log({ ...settings, [field]: "en" });

console.log(Object.fromEntries([["theme", "light"], ["fontSize", 14]]));

Object.freeze(settings);
console.log(Object.isFrozen(settings));
```

### LESSON 28 — Mini challenge

```typescript
function countValues(obj) {
  let total = 0;

  for (const value of Object.values(obj)) {
    if (value !== undefined && value !== null) {
      total = total + 1;
    }
  }

  return total;
}

console.log(countValues({ a: 1, b: null, c: undefined, d: 0 }));

function invert(obj) {
  const result = {};

  for (const [key, value] of Object.entries(obj)) {
    result[value] = key;
  }

  return result;
}

console.log(invert({ a: 1, b: 2 }));

const original = { name: "Mia", address: { city: "Rome" } };
const copy = { ...original };
copy.address.city = "Milan";

console.log(original.address.city);
// Spread copies one level only, so both objects point at the SAME address object.

const deepOriginal = { name: "Mia", address: { city: "Rome" } };
const deepCopy = structuredClone(deepOriginal);
deepCopy.address.city = "Milan";

console.log(deepOriginal.address.city, deepCopy.address.city);
// structuredClone copies every level, so the two address objects are now separate.

function createCounter() {
  let count = 0;

  return {
    increment() {
      count = count + 1;
    },
    get value() {
      return count;
    },
  };
}

const counter = createCounter();

counter.increment();
counter.increment();

console.log(counter.value);
console.log(Object.keys(counter));   // no "count" key: it is a variable, not a property
```
