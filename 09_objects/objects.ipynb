# JavaScript — Objects

## LESSON 27 — Objects

An **object** groups related values under names instead of positions.

```js
const user = {
  name: "Sam",
  age: 34,
  isAdmin: false,
};
```

Each entry is a **key: value** pair. An array asks *"which position?"*, an object asks *"which name?"*.

### Reading values

```js
user.name;      // "Sam"      -> dot notation
user["name"];   // "Sam"      -> bracket notation

const key = "age";
user[key];      // 34         -> only brackets work here
```

Dot notation is shorter, so it is what you write most of the time. Brackets are not just a second style — there are two cases where they are the only thing that works.

**1. The key is inside a variable.** `user.key` looks for a key literally named `"key"`. `user[key]` reads the variable first, gets `"age"`, and looks for that.

**2. The key is not a plain word.** A space, a dash or a leading digit cannot follow a dot.

```js
const config = { "max-size": 10, "first name": "Sam", 2020: "leap" };

config["max-size"];     // 10
config["first name"];   // "Sam"
config[2020];           // "leap"
```

That last line has a number in the brackets, but the key itself is the **string** `"2020"`. Object keys are always strings — JavaScript converts anything else on the way in.

### Adding, changing, deleting

```js
user.city = "Geneva";   // add a new key
user.age = 35;          // change an existing one
delete user.isAdmin;    // remove it
```

### Shorthand, when the variable already has the right name

If the key and the variable have the same name, write it once.

```js
const name = "Sam";
const age = 34;

const user = { name, age };   // { name: "Sam", age: 34 }
```

`{ name, age }` is only a shorter way of typing `{ name: name, age: age }` — nothing else changes.

### Nested objects

```js
const product = {
  name: "Laptop",
  specs: { ram: "16GB", cpu: "M3" },
};

product.specs.ram;   // "16GB"
```

A missing key and a missing **level** are two very different things:

```js
product.price;          // undefined   -> no such key, no error
product.specs.weight;   // undefined   -> same thing, one level down
product.seller.name;    // TypeError: Cannot read properties of undefined (reading 'name')
```

The third line fails because `product.seller` is `undefined`, and `undefined` has no properties to read. **Reading a missing key is safe; reading *through* one is not.** For now, check the level before you go deeper — LESSON 32 has a one-character tool for it, `?.`.

### Methods

A value can be a function. Then it's called a **method**.

```js
const counter = {
  total: 0,
  increment() {
    this.total = this.total + 1;
  },
};

counter.increment();
```

`this` is the object **in front of the dot at the moment of the call** — `counter`, on that last line. It is decided when the method runs, not where it is written.

Write methods with the shorthand above. An arrow function has no `this` of its own, so it never works as a method:

```js
const broken = {
  total: 0,
  increment: () => {
    this.total = this.total + 1;   // `this` is not the object here
  },
};
```

### Listing keys and values

```js
Object.keys(user);          // ["name", "age", ...]
Object.values(user);        // ["Sam", 34, ...]
Object.keys(user).length;   // how many entries the object holds
```

`Object.keys` hands you an array, so `for...of` from LESSON 17 can walk it. `for...in` from LESSON 18 walks the keys directly:

```js
for (const key in user) {
  console.log(key, user[key]);
}
```

Note the `user[key]`: inside the loop the key lives in a variable, which is case 1 from above.

### Two names, one object

An object is stored by **reference**. The variable holds a pointer to the object, not the values themselves. That has two visible consequences:

```js
const a = { count: 1 };
const b = a;       // not a copy: the SAME object, with a second name

b.count = 99;
a.count;           // 99   <- changed through the other name

a === b;                 // true    -> one object, two names
a === { count: 99 };     // false   -> same content, different object
```

`===` on objects asks *"is it the same object?"*, never *"do they hold the same thing?"*. Two objects written out separately are always different objects, however identical they look. Copying one properly is the first thing LESSON 28 deals with.

The same reasoning explains `const`: it forbids pointing the variable at a *different* object, not changing the one it already points at — exactly like the arrays in LESSON 25.

### Key notes

- Use **dot notation** when you know the key while writing the code. Use **brackets** when the key is in a variable, or when it has a space, a dash or a leading digit — `user.key` looks for a key literally named `"key"`.
- **Every key is a string.** `{ 2020: "leap" }` stores `"2020"`, and `Object.keys` gives it back as text.
- Reading a key that doesn't exist gives `undefined`, it does **not** throw an error. Reading **through** a missing key — `product.seller.name` — does throw.
- Inside a method, `this` is the object in front of the dot **at call time**. Arrow functions have no `this` of their own, so never use one as a method.
- **Objects are compared by identity.** `{ a: 1 } === { a: 1 }` is `false`, and `const b = a` gives one object two names, not two objects.

### Example

````typescript
const exampleUser = {
  name: "Sam",
  age: 34,
  isAdmin: false,
};

console.log(exampleUser.name);

const exampleKey = "age";
console.log(exampleUser[exampleKey]);
console.log(exampleUser.exampleKey);   // no key with that literal name

// keys a dot cannot reach, and keys that are always strings
const exampleConfig = { "max-size": 10, "first name": "Sam", 2020: "leap" };
console.log(exampleConfig["max-size"], exampleConfig["first name"], exampleConfig[2020]);
console.log(Object.keys(exampleConfig));   // "2020" first: LESSON 28 explains the order

// shorthand: same name on both sides, written once
{
  const name = "Mia";
  const age = 28;
  console.log({ name, age });
}

exampleUser.city = "Geneva";
exampleUser.age = 35;
delete exampleUser.isAdmin;
console.log(exampleUser);

console.log(exampleUser.phone);   // undefined, not an error

// a missing key is safe, a missing level is not
const exampleProduct = { name: "Laptop", specs: { ram: "16GB", cpu: "M3" } };
console.log(exampleProduct.specs.ram);
console.log(exampleProduct.specs.weight);
// exampleProduct.seller.name;   // uncomment to see the TypeError

console.log(Object.keys(exampleUser));
console.log(Object.values(exampleUser));
console.log(Object.keys(exampleUser).length);

for (const key in exampleUser) {
  console.log(key, exampleUser[key]);
}

const exampleCounter = {
  total: 0,
  increment() {
    this.total = this.total + 1;
  },
};

exampleCounter.increment();
exampleCounter.increment();
console.log(exampleCounter.total);

// two names, one object
const exampleFirst = { count: 1 };
const exampleSecond = exampleFirst;
exampleSecond.count = 99;
console.log(exampleFirst.count);
console.log({ count: 1 } === { count: 1 });
````

## LESSON 28 — Objects in depth

LESSON 27 covered one key at a time. This lesson is about the whole object: walking it, copying it, asking what is inside it, and deciding what the outside world is allowed to touch.

### Reading the whole object

Three functions turn an object into an array you can loop over.

```js
const user = { name: "Mia", age: 28 };

Object.keys(user);      // ["name", "age"]
Object.values(user);    // ["Mia", 28]
Object.entries(user);   // [["name", "Mia"], ["age", 28]]
```

`entries` is the one that gives you both halves at once. Each entry is a **two-item array**: the key at index 0, the value at index 1. So this works, and it is what you would write with only LESSON 27 in hand:

```js
for (const entry of Object.entries(user)) {
  console.log(entry[0], entry[1]);   // "name" "Mia", then "age" 28
}
```

`entry[0]` and `entry[1]` are hard to read, so JavaScript lets you name the two slots right in the `for` line:

```js
for (const [key, value] of Object.entries(user)) {
  console.log(`${key}: ${value}`);
}
```

The square brackets on the **left** are not an array — they are a pattern. They say *"this is a two-item array; call item 0 `key` and item 1 `value`"*. That move is called **destructuring**, and LESSON 31 covers it properly. For now, read `const [key, value]` as "unpack the pair".

### From a list of pairs back to an object

`Object.fromEntries` is the return trip: hand it a list of `[key, value]` pairs and it builds the object.

```js
Object.fromEntries([["name", "Mia"], ["age", 28]]);   // { name: "Mia", age: 28 }

Object.fromEntries(Object.entries(user));             // a copy of user, the long way round
```

Whenever you need to rebuild an object with different keys or different values, the pattern is the same: `entries` on the way out, change the pairs, `fromEntries` on the way back in.

### Copying an object

```js
const defaults = { theme: "light", size: "m" };

const settings = { ...defaults, size: "l" };   // { theme: "light", size: "l" }
```

The `...` reads every key/value pair out of `defaults` and drops them into the new object, and then `size: "l"` is written on top. **Later keys win** — that is the whole rule. It is the same `...` you saw on arrays in LESSON 26, and LESSON 31 covers it properly.

`Object.assign` does the same job with older syntax, and you will meet it in existing code:

```js
Object.assign({}, defaults, { size: "l" });   // { theme: "light", size: "l" }
```

| | what it does |
|---|---|
| `{ ...a, ...b }` | a new object with `a`'s pairs, then `b`'s on top |
| `Object.assign({}, a, b)` | the same result, older spelling |
| `Object.assign(a, b)` | **no new object** — `b` is poured into `a`, and `a` changes |

That third line is the one to watch: the first argument is the target, so passing a real object instead of `{}` modifies it.

### How deep does the copy go?

This is the trap of the lesson, and it costs more time than any other object mistake.

Spread and `Object.assign` copy **one level**. A nested object is not copied — only the reference to it is, exactly as in LESSON 27. So the "copy" and the original end up sharing it:

```js
const original = { name: "Mia", address: { city: "Rome" } };
const copy = { ...original };

copy.name = "Sam";             // a string, copied -> only the copy changes
copy.address.city = "Milan";   // an object, shared -> both change

original.name;           // "Mia"     -> fine
original.address.city;   // "Milan"   <- the original changed too
```

Nothing warns you. The copy looks independent until the day a nested value moves in both objects at once.

The fix is `structuredClone`, which copies every level:

```js
const safe = structuredClone(original);

safe.address.city = "Turin";
original.address.city;   // "Milan"   -> untouched this time
```

Two limits worth knowing before you reach for it: `structuredClone` cannot copy an object that holds a **function** — it throws `DataCloneError` — and it cannot copy a DOM node.

| | nested objects are |
|---|---|
| `const b = a` | not copied at all — one object, two names |
| `{ ...a }` / `Object.assign({}, a)` | **shared** with the original |
| `structuredClone(a)` | copied, all the way down |

Rule of thumb: **flat object, spread is enough. Nested object you intend to change, `structuredClone`.**

### Does this key exist?

Three ways to ask, and they do not answer the same question.

```js
const user = { name: "Mia", age: 28, nickname: undefined };

Object.hasOwn(user, "name");   // true
"name" in user;                // true
user.name !== undefined;       // true
```

The differences show up at the edges:

```js
Object.hasOwn(user, "nickname");   // true    -> the key is there
user.nickname !== undefined;       // false   <- lies: the key exists, the value is undefined

Object.hasOwn(user, "toString");   // false   -> not a key of yours
"toString" in user;                // true    <- every object inherits toString
```

| | finds a key holding `undefined` | finds inherited keys |
|---|---|---|
| `Object.hasOwn(obj, key)` | yes | no |
| `key in obj` | yes | **yes** |
| `obj.key !== undefined` | **no** | yes |

`Object.hasOwn` is the one to reach for: it answers "does *this* object have this key", which is what you almost always mean. Use `in` only when inherited keys should count too.

(Older code writes the same check as `obj.hasOwnProperty(key)`. `Object.hasOwn` replaced it in 2022 and is safer — recognize the old one, write the new one.)

### Computed keys

Square brackets inside the literal let a variable supply the key.

```js
const field = "email";
const record = { [field]: "sam@example.com" };   // { email: "sam@example.com" }
```

Without the brackets you would get a key literally named `"field"` — the same dot-versus-bracket rule from LESSON 27, this time on the writing side.

### The order of the keys

`Object.keys`, `Object.values`, `Object.entries` and `for...in` all walk the object in the same order, and it is not always the order you typed:

```js
Object.keys({ b: 1, 2: "two", a: 3, 1: "one" });   // ["1", "2", "b", "a"]
```

Keys that look like whole numbers come **first, in numeric order**. Everything else follows in insertion order. It rarely matters, but when a numeric-looking key jumps to the front, this is why. If the order is part of your data, keep it in an array — or use a `Map`, in LESSON 35.

### Getters and setters

A **getter** runs like a function but is read like a value. A **setter** runs when you assign to it.

```js
const account = {
  balance: 100,

  get formatted() {
    return `${this.balance} EUR`;
  },

  set deposit(amount) {
    this.balance += amount;
  },
};

account.formatted;      // "100 EUR"   -> read WITHOUT parentheses
account.deposit = 50;   // written like a property, runs like a function
account.formatted;      // "150 EUR"
```

Two rules: a getter must `return` something, or reading it gives `undefined`; a setter takes **exactly one** parameter, the value being assigned.

### Keeping data private

By default an object hides nothing: anything holding a reference to it can read, change or delete any key. There are three levels of protection, and it is worth knowing which one is real.

**1. The `_` convention — no protection at all.** A leading underscore is a note to other programmers saying "this is internal, don't touch". JavaScript does not enforce it.

```js
const timer = { _startedAt: 0 };

timer._startedAt = 99;   // works fine, nothing stops you
```

**2. `Object.freeze` — no changes, but no secrets either.** A frozen object refuses new keys, changes and deletions. Everything stays readable.

```js
const config = Object.freeze({ id: 1, tags: ["a"] });

config.id;                 // 1
Object.isFrozen(config);   // true

config.id = 99;            // TypeError: Cannot assign to read only property 'id'
delete config.id;          // TypeError: Cannot delete property 'id'
```

Two things to remember. **What that last pair of lines does depends on the mode the code is running in.** Notebook cells, modules and class bodies all run in *strict mode*, where a write to a frozen object throws the `TypeError` above. Old scripts run in *sloppy mode*, where the very same line does nothing at all and says nothing about it. Everything you write today is strict, so expect the error — but when you read old code, remember the silent version exists.

The other thing: freeze is **shallow**, exactly like spread. A nested object inside a frozen object is still wide open:

```js
config.tags.push("b");   // works! the array itself was never frozen
config.tags;             // ["a", "b"]
```

**3. A factory function — actually private.** Declare the value as a normal variable inside a function, and return an object whose methods use it. Nothing outside that function has a name for the variable, so nothing outside can reach it.

```js
function createAccount(startingBalance) {
  let balance = startingBalance;   // a plain variable, not a key

  return {
    deposit(amount) {
      balance += amount;
    },
    get balance() {
      return balance;
    },
  };
}

const account = createAccount(100);

account.deposit(50);
account.balance;        // 150
Object.keys(account);   // ["deposit", "balance"]   -> the variable is nowhere in the object

account.balance = 0;    // TypeError: Cannot set property balance of #<Object> which has only a getter
```

`balance` can be read and can be changed through `deposit`, and there is no third way in. The object keeps the variable alive because its methods still use it — a function that outlives its call like this is called a **closure**, and it is what makes the value genuinely private.

That is how a plain object gets privacy. Objects built from a `class` have private syntax of their own, `#`, in LESSON 33.

### Key notes

- **`entries` gives pairs, not values.** `for (const [key, value] of Object.entries(obj))` — if you get an array where you expected a string, the brackets on the left are missing.
- **Spread and `Object.assign` copy one level deep.** Change a nested object in the copy and the original changes too. Reach for `structuredClone(obj)` when the object has levels you intend to change.
- **`Object.assign(a, b)` modifies `a`.** Pass `{}` as the first argument unless you really mean to change the target.
- **`Object.hasOwn(obj, key)` is the key check to default to.** `in` also finds inherited keys like `toString`; `obj.key !== undefined` cannot tell a missing key from one holding `undefined`.
- `Object.keys` returns **strings**, always, and puts whole-number keys first.
- A getter is read **without parentheses** and must `return`. A setter takes exactly one parameter.
- **`Object.freeze` is shallow, and only stops the top level.** In a notebook cell or a module, writing to a frozen key throws a `TypeError`; in an old sloppy-mode script the same line is ignored without a word. Either way, a nested object inside it can still be changed.
- **Real privacy comes from a variable nothing else can name** — a factory function returning an object. An underscore is only a request.

### Example

````typescript
const exampleProfile = { name: "Mia", age: 28, city: "Rome" };

console.log(Object.keys(exampleProfile));
console.log(Object.values(exampleProfile));
console.log(Object.entries(exampleProfile));

// one entry is a two-item array: index 0 is the key, index 1 is the value
for (const entry of Object.entries(exampleProfile)) {
  console.log(entry[0], entry[1]);
}

// the same loop, with the pair unpacked into two names
for (const [key, value] of Object.entries(exampleProfile)) {
  console.log(`${key}: ${value}`);
}

// pairs back into an object
console.log(Object.fromEntries([["name", "Mia"], ["age", 28]]));

// copying: later keys win
const exampleDefaults = { theme: "light", size: "m" };
console.log({ ...exampleDefaults, size: "l" });
console.log(Object.assign({}, exampleDefaults, { theme: "dark" }));
console.log(exampleDefaults);   // untouched: {} was the target

// how deep does the copy go?
const exampleOriginal = { name: "Mia", address: { city: "Rome" } };
const exampleCopy = { ...exampleOriginal };

exampleCopy.name = "Sam";             // a string: only the copy changes
exampleCopy.address.city = "Milan";   // an object: shared, so both change

console.log(exampleOriginal.name);
console.log(exampleOriginal.address.city);

const exampleSafe = structuredClone(exampleOriginal);
exampleSafe.address.city = "Turin";
console.log(exampleOriginal.address.city, exampleSafe.address.city);

// does this key exist?
const exampleUser2 = { name: "Mia", nickname: undefined };

console.log(Object.hasOwn(exampleUser2, "name"), "name" in exampleUser2);
console.log(Object.hasOwn(exampleUser2, "nickname"), exampleUser2.nickname !== undefined);
console.log(Object.hasOwn(exampleUser2, "toString"), "toString" in exampleUser2);

// a key supplied by a variable
const exampleField = "email";
console.log({ [exampleField]: "sam@example.com" });

// whole-number keys come first
console.log(Object.keys({ b: 1, 2: "two", a: 3, 1: "one" }));

// getters and setters
const exampleAccount = {
  balance: 100,
  get formatted() {
    return `${this.balance} EUR`;
  },
  set deposit(amount) {
    this.balance += amount;
  },
};

console.log(exampleAccount.formatted);
exampleAccount.deposit = 50;
console.log(exampleAccount.formatted);

// freeze blocks the top level only, and throws here because notebook cells are strict
const exampleFrozen = Object.freeze({ id: 1, tags: ["a"] });
console.log(exampleFrozen.id, Object.isFrozen(exampleFrozen));
// exampleFrozen.id = 99;   // uncomment to see the TypeError
exampleFrozen.tags.push("b");
console.log(exampleFrozen.tags);   // the nested array was never frozen

// a factory function: balance is a variable, not a key
function createExampleAccount(startingBalance) {
  let balance = startingBalance;

  return {
    deposit(amount) {
      balance += amount;
    },
    get balance() {
      return balance;
    },
  };
}

const examplePrivate = createExampleAccount(100);
examplePrivate.deposit(50);
console.log(examplePrivate.balance);
console.log(Object.keys(examplePrivate));
// examplePrivate.balance = 0;   // uncomment: TypeError, there is no setter

// a function inside the object is what structuredClone cannot copy
// structuredClone(examplePrivate);   // uncomment to see the DataCloneError
````
