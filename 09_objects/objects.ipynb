# JavaScript — Objects

## LESSON 27 — Objects

An **object** groups related values under names instead of positions.

```js
const user = {
  name: "Sam",
  age: 34,
  isAdmin: false,
};
```

Each entry is a **key: value** pair. An array asks *"which position?"*, an object asks *"which name?"*.

### Reading values

```js
user.name;      // "Sam"      -> dot notation
user["name"];   // "Sam"      -> bracket notation

const key = "age";
user[key];      // 34         -> only brackets work here
```

### Adding, changing, deleting

```js
user.city = "Geneva";   // add a new key
user.age = 35;          // change an existing one
delete user.isAdmin;    // remove it
```

### Nested objects

```js
const product = {
  name: "Laptop",
  specs: { ram: "16GB", cpu: "M3" },
};

product.specs.ram;   // "16GB"
```

### Methods

A value can be a function. Then it's called a **method**.

```js
const counter = {
  total: 0,
  increment() {
    this.total = this.total + 1;
  },
};

counter.increment();
```

### Listing keys and values

```js
Object.keys(user);     // ["name", "age", ...]
Object.values(user);   // ["Sam", 34, ...]
```

### Key notes

- Use **dot notation** when you know the key while writing the code. Use **brackets** when the key is inside a variable — `user.key` looks for a key literally named `"key"`.
- Reading a key that doesn't exist gives `undefined`, it does **not** throw an error.
- Inside a method, `this` refers to the object the method was called on.

### Example

````typescript
const exampleUser = {
  name: "Sam",
  age: 34,
  isAdmin: false,
};

console.log(exampleUser.name);

const exampleKey = "age";
console.log(exampleUser[exampleKey]);

exampleUser.city = "Geneva";
delete exampleUser.isAdmin;
console.log(exampleUser);

console.log(exampleUser.phone);   // undefined, not an error

console.log(Object.keys(exampleUser));
console.log(Object.values(exampleUser));

const exampleCounter = {
  total: 0,
  increment() {
    this.total = this.total + 1;
  },
};

exampleCounter.increment();
exampleCounter.increment();
console.log(exampleCounter.total);
````

## LESSON 28 — Objects in depth

### Reading the whole object

```js
const user = { name: "Mia", age: 28 };

Object.keys(user);       // ["name", "age"]
Object.values(user);     // ["Mia", 28]
Object.entries(user);    // [["name", "Mia"], ["age", 28]]
```

`entries` is the one that unlocks looping with both parts at once:

```js
for (const [key, value] of Object.entries(user)) {
  console.log(`${key}: ${value}`);
}
```

### Copying and merging

```js
Object.assign({}, defaults, overrides);    // merge into a new object
{ ...defaults, ...overrides };             // the same thing, shorter
```

Later keys win. Both make a **shallow** copy: nested objects are still shared with the original.

### Checking a key exists

```js
"name" in user;            // true
user.age !== undefined;    // also true, but lies when the value IS undefined
```

Use `in` when you need to know whether the key is there, regardless of its value.

### Computed keys

Square brackets in a literal let you use a variable as the key.

```js
const field = "email";
const record = { [field]: "sam@example.com" };    // { email: "..." }
```

### Getters, setters and freezing

```js
const account = {
  balance: 100,
  get formatted() {
    return `${this.balance} EUR`;
  },
  set deposit(amount) {
    this.balance += amount;
  },
};

account.formatted;      // "100 EUR"   <- read like a property
account.deposit = 50;   // written like a property, runs like a function
```

`Object.freeze(obj)` blocks any further change. Assignments then fail silently, so it is a guard for constants rather than a security measure.

### Key notes

- **Spread and `Object.assign` copy one level deep.** Change a nested object in the copy and the original changes too.
- **`in` beats `!== undefined`** when a key may legitimately hold `undefined`.
- `Object.keys` returns **strings**, always — even for numeric-looking keys.
- `Object.freeze` fails **silently** in normal code. Nothing throws; the assignment just does not happen.

### Example

````typescript
const exampleUser = { name: "Mia", age: 28, city: "Rome" };

console.log(Object.keys(exampleUser));
console.log(Object.values(exampleUser));
console.log(Object.entries(exampleUser));

for (const [key, value] of Object.entries(exampleUser)) {
  console.log(`${key}: ${value}`);
}

const exampleDefaults = { theme: "light", size: "m" };
console.log({ ...exampleDefaults, size: "l" });
console.log(Object.assign({}, exampleDefaults, { theme: "dark" }));

console.log("name" in exampleUser, "email" in exampleUser);

const exampleField = "email";
console.log({ [exampleField]: "sam@example.com" });

const exampleAccount = {
  balance: 100,
  get formatted() {
    return `${this.balance} EUR`;
  },
  set deposit(amount) {
    this.balance += amount;
  },
};

console.log(exampleAccount.formatted);
exampleAccount.deposit = 50;
console.log(exampleAccount.formatted);

const exampleFrozen = Object.freeze({ id: 1 });
exampleFrozen.id = 99;
console.log(exampleFrozen.id);
````
