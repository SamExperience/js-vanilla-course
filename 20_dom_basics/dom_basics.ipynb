## LESSON 44 — DOM basics

The **DOM** is the page, seen as objects JavaScript can read and change.

### Finding elements

```js
document.querySelector("#count");     // ONE element, by CSS selector
document.querySelector(".card");      // the FIRST match
document.querySelectorAll("li");      // ALL of them, as a list
```

The argument is a CSS selector — the same string you would write in a stylesheet. `#` for an id, `.` for a class.

### Changing content

```js
element.textContent = "Hello";              // sets TEXT
element.innerHTML = "<em>Hello</em>";       // parses HTML tags
```

Give both the same string, and the difference is obvious:

```js
element.textContent = "<em>Hello</em>";     // shows: <em>Hello</em>
element.innerHTML = "<em>Hello</em>";       // shows: Hello, in italics
```

### Changing appearance

```js
element.classList.add("highlight");
element.classList.remove("highlight");
element.classList.toggle("highlight");   // adds it if absent, removes it if present

element.style.color = "red";             // inline style, for one-offs
```

### Creating elements

```js
const list = document.querySelector("#fruits");   // where it will go

const item = document.createElement("li");
item.textContent = "Apple";
list.append(item);                       // nothing appears until you append
```

### Key notes

- **`textContent` vs `innerHTML`.** `textContent` puts characters on the page, exactly as you wrote them. `innerHTML` hands your string to the HTML parser, which turns tags into real elements. Default to `textContent`: it is faster, and it never turns text into live markup behind your back. That matters as soon as the text is not yours — a username, a comment, anything typed by someone else. Reach for `innerHTML` only when you genuinely mean tags.
- **`classList` or `style`?** A class keeps the look in the stylesheet, where it belongs, and lets you change it in one place. `style` writes the rule straight onto the element, which normally beats what the stylesheet says. Reach for `style` when a single value belongs to that one element and nothing else; reach for `classList` the rest of the time.
- `querySelector` returns `null` when nothing matches — it does **not** throw. The error `Cannot set properties of null` therefore arrives one line *later*, when you try to use what you never got. It always means your selector found nothing: check the `#`, the `.`, and the spelling. When in doubt, `console.log` the result before you use it.
- `querySelectorAll` gives you a **NodeList**, not an array. `.length` and `.forEach` work, which covers most of what you want, but `.map`, `.filter` and `.find` are not there. Spread copies it into a real array when you need them — the same trick you used on a `Set`: `[...document.querySelectorAll("li")]`.
- A created element exists only in memory until you `append` it somewhere.
- Prefer `classList` over rewriting `className`: `className = "x"` wipes every other class.

### The pattern behind every list on screen — runnable

Building UI from data is really just a loop. The cell below runs that loop — the part
that has nothing to do with the browser. What changes in `project/` is only the body:
instead of building an object, each pass creates a real `<li>`, sets its text, and
appends it.

````typescript
const fruits = ["Apple", "Banana", "Orange"];

// In the browser, each pass of this loop does three things:
//   create an element  ->  put text in it  ->  append it to a parent.
// There is no document here, so a plain object stands in for the element.
const list = [];

fruits.forEach((fruit) => {
  const item = { tag: "li", textContent: fruit };  // document.createElement("li")
  list.push(item);                                 // list.append(item)
});

console.log(list.length, "items built");
console.log(list);
````
