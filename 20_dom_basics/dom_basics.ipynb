## LESSON 44 — DOM basics

The **DOM** is the page, seen as objects JavaScript can read and change.

### Finding elements

```js
document.querySelector("#count");     // ONE element, by CSS selector
document.querySelector(".card");      // the FIRST match
document.querySelectorAll("li");      // ALL of them, as a list
```

The argument is a CSS selector — the same string you'd write in a stylesheet. `#` for an id, `.` for a class.

### Changing content

```js
element.textContent = "Hello";              // sets TEXT
element.innerHTML = "<em>Hello</em>";       // parses HTML tags
```

### Changing appearance

```js
element.classList.add("highlight");
element.classList.remove("highlight");
element.classList.toggle("highlight");   // returns true if now present

element.style.color = "red";             // inline style, for one-offs
```

### Creating elements

```js
const item = document.createElement("li");
item.textContent = "Apple";
list.append(item);                       // nothing appears until you append
```

### Key notes

- **`textContent` vs `innerHTML`.** Use `textContent` for text — it's faster and safe. Reach for `innerHTML` only when you genuinely mean tags.
- `querySelector` returns `null` when nothing matches. The error `Cannot set properties of null` always means your selector found nothing — check the `#` or the spelling.
- A created element exists only in memory until you `append` it somewhere.
- Prefer `classList` over rewriting `className`: `className = "x"` wipes every other class.

### The pattern behind every list on screen — runnable

Building UI from data is really just a loop. The cell below does it with strings,
so it runs here without a browser. In `project/` the very same loop creates real
`<li>` elements instead.

````typescript
const fruits = ["Apple", "Banana", "Orange"];

// In the browser: create an <li> per fruit and append it.
// Here: build the same thing as text, just to see the shape.
fruits.forEach((fruit) => {
  console.log(`<li>${fruit}</li>`);
});
````
