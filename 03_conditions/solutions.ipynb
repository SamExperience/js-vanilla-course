### LESSON 11 — Mini challenge

```typescript
let userAge = 25;

if (userAge >= 18) {
  console.log("Adult");
} else if (userAge >= 13) {
  console.log("Teenager");
} else console.log("Child");
```

### LESSON 12 — Exercise

```typescript
let selectedRole = "admin";

switch (selectedRole) {
  case "admin":
    console.log("Full access");
    break;
  case "editor":
    console.log("Edit access");
    break;
  case "user":
    console.log("Read access");
    break;
  default:
    console.log("No access");
}
```

### LESSON 12 — Mini challenge

```typescript
let menuOption = 2;

switch (menuOption) {
  case 1:
    console.log("Home");
    break;
  case 2:
    console.log("Profile");
    break;
  case 3:
    console.log("Settings");
    break;
  default:
    console.log("Invalid option");
}
```
