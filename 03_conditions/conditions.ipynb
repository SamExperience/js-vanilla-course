## LESSON 10 — if / else

- `if` → execute code when the condition is true.
- `else` → execute code when the condition is false.

```js
if (condition) {
    // code
} else {
    // code
}
```

### Key notes

The condition must produce a boolean.

```js
age >= 18   // -> true / false
```

- `if (true)` → execute
- `if (false)` → skip

### Example

````typescript
const exampleAge = 20;

if (exampleAge >= 18) {
  console.log("Adult");
} else {
  console.log("Minor");
}

const exampleLoggedIn = true;

if (exampleLoggedIn) {
  console.log("Welcome");
}
````

## LESSON 11 — else if

`else if` allows you to test multiple conditions. JavaScript checks them from top to bottom and **stops at the first one that is true**.

```js
if (condition1) {
    // case 1
} else if (condition2) {
    // case 2
} else {
    // fallback
}
```

### Key notes

- `if` → first condition
- `else if` → another condition
- `else` → runs if all previous conditions are false

**Order matters:**

```js
if (score >= 60) {
    ...
} else if (score >= 80) {   // never reached!
    ...
}
```

The second condition can never be reached, because 80 is already `>= 60`.

### Example

````typescript
const exampleScore = 75;

if (exampleScore >= 80) {
  console.log("Excellent");
} else if (exampleScore >= 60) {
  console.log("Good");
} else {
  console.log("Needs improvement");
}
````

## LESSON 12 — switch

`switch` is useful when you compare ONE value against multiple **exact** values.

```js
switch (value) {
  case value1:
    // code
    break;

  case value2:
    // code
    break;

  default:
    // fallback
}
```

### Key notes

- `case` → checks for an exact match
- `break` → stops the switch
- `default` → runs when no case matches

`switch` is mainly useful for exact values, not for ranges.

### Example

````typescript
const exampleDay = "Monday";

switch (exampleDay) {
  case "Monday":
    console.log("Start of the week");
    break;

  case "Friday":
    console.log("Almost weekend");
    break;

  case "Sunday":
    console.log("Weekend");
    break;

  default:
    console.log("Normal day");
}
````

## LESSON 13 — Ternary operator

A short way to write a simple `if / else`.

```js
condition ? valueIfTrue : valueIfFalse;
```

```js
const age = 20;
const result = age >= 18 ? "Adult" : "Minor";
```

### Key notes

- `?` → value when the condition is true
- `:` → value when the condition is false

Use a ternary for **simple** decisions. Prefer `if / else` when the logic gets complex.

### Example

````typescript
const exampleIsLoggedIn = true;

const exampleMessage = exampleIsLoggedIn ? "Welcome" : "Please log in";

console.log(exampleMessage);
````
