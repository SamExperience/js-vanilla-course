### LESSON 42 — Exercise

```typescript
const delay = (ms) => new Promise((resolve) => setTimeout(resolve, ms));

async function getProduct() {
  await delay(200);
  return { name: "Laptop", price: 1200 };
}

const product = await getProduct();
console.log(product.name);

// Without await you get the Promise itself, not the product
console.log(getProduct());
```

### LESSON 42 — Mini challenge

```typescript
async function login(username) {
  await delay(200);

  if (username !== "sam") {
    throw new Error("unknown user");
  }

  return "Welcome, sam";
}

try {
  const message = await login("sam");
  console.log(message);
} catch (error) {
  console.log("Caught:", error.message);
}

try {
  const message = await login("alex");
  console.log(message);
} catch (error) {
  console.log("Caught:", error.message);
}
```

### LESSON 43 — Exercise

```typescript
function delay(ms, value) {
  return new Promise((resolve) => setTimeout(() => resolve(value), ms));
}

delay(100, "hello").then((value) => console.log("then:", value));

console.log("await:", await delay(100, "hello"));

const started = Date.now();
const [a, b] = await Promise.all([delay(100, "a"), delay(100, "b")]);
console.log(a, b);
console.log(`took about ${Date.now() - started}ms — not 200`);

console.log(await Promise.race([delay(20, "fast"), delay(200, "slow")]));
```

### LESSON 43 — Mini challenge

```typescript
function unreliable(shouldFail) {
  return new Promise((resolve, reject) => {
    setTimeout(() => {
      if (shouldFail) reject(new Error("failed"));
      else resolve("ok");
    }, 20);
  });
}

const settled = await Promise.allSettled([
  unreliable(false),
  unreliable(true),
  unreliable(false),
]);

console.log(settled.map((result) => result.status));

try {
  await Promise.all([unreliable(false), unreliable(true), unreliable(false)]);
} catch (error) {
  console.log("all rejected:", error.message);
  console.log("the two successful results are gone — that is the trade-off");
}

let count = 0;
const ticker = setInterval(() => {
  count += 1;
  console.log(count);
  if (count === 3) clearInterval(ticker);
}, 20);
```
