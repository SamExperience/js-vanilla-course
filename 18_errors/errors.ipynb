# JavaScript — Errors

## LESSON 41 — Errors

You met `try`/`catch` in LESSON 42's neighbourhood already, wrapped around `await`. This lesson is about the other half: deciding when **your own** code should refuse to continue.

### throw

```js
function withdraw(balance, amount) {
  if (amount > balance) {
    throw new Error("Insufficient funds");
  }
  return balance - amount;
}
```

`throw` stops the function immediately. Nothing after it runs, and the error travels up to whoever called you — until something catches it, or the program stops.

Returning `null` to signal a problem forces every caller to remember to check. Throwing makes the problem impossible to ignore. Throw when continuing would be wrong; return a value when "nothing found" is a normal outcome.

### The Error object

```js
const error = new Error("Insufficient funds");

error.message;    // "Insufficient funds"
error.name;       // "Error"
error.stack;      // where it happened
```

### Your own error types

A custom class lets a caller react to *this* problem specifically instead of to any problem at all.

```js
class ValidationError extends Error {
  constructor(field) {
    super(`${field} is required`);
    this.name = "ValidationError";
    this.field = field;
  }
}

try {
  throw new ValidationError("email");
} catch (error) {
  if (error instanceof ValidationError) {
    console.log(error.field);      // "email"
  } else {
    throw error;                   // not mine — pass it on
  }
}
```

### finally

`finally` runs either way: after a successful `try`, and after a `catch`. It is where cleanup goes — hiding a spinner, closing something you opened.

```js
try {
  render(await load());
} catch (error) {
  showError(error.message);
} finally {
  hideSpinner();            // always
}
```

### Key notes

- **Throw the `Error` object, not a string.** `throw "oops"` runs happily. What you catch then has no `message` and no stack, so every `error.message` in your code quietly becomes `undefined`.
- **An empty `catch` is a bug in waiting.** Swallowing an error means the failure happens anyway, silently, somewhere later.
- **Only catch what you can handle.** Re-`throw` anything else, so a real bug does not get mistaken for an expected failure.
- `finally` runs even when the `try` block returns. That is exactly why cleanup belongs there and not after the block.

### Example

````typescript
function exampleWithdraw(balance, amount) {
  if (typeof amount !== "number") {
    throw new TypeError("amount must be a number");
  }
  if (amount > balance) {
    throw new Error("Insufficient funds");
  }
  return balance - amount;
}

class ExampleValidationError extends Error {
  constructor(field) {
    super(`${field} is required`);
    this.name = "ExampleValidationError";
    this.field = field;
  }
}

try {
  console.log(exampleWithdraw(100, 30));
  console.log(exampleWithdraw(100, 500));
} catch (error) {
  console.log(error.name, "->", error.message);
} finally {
  console.log("done either way");
}

try {
  throw new ExampleValidationError("email");
} catch (error) {
  if (error instanceof ExampleValidationError) {
    console.log("missing field:", error.field);
  } else {
    throw error;
  }
}
````
