### LESSON 41 — Exercise

```typescript
function divide(a, b) {
  if (b === 0) {
    throw new Error("Cannot divide by zero");
  }
  return a / b;
}

for (const [a, b] of [[10, 2], [10, 0]]) {
  try {
    console.log(divide(a, b));
  } catch (error) {
    console.log(error.name, "->", error.message);
  } finally {
    console.log("attempt finished");
  }
}
```

### LESSON 41 — Mini challenge

```typescript
class NotFoundError extends Error {
  constructor(id) {
    super(`No user with id ${id}`);
    this.name = "NotFoundError";
    this.id = id;
  }
}

const users = [{ id: 1, name: "Mia" }, { id: 2, name: "Alex" }];

function findUser(id) {
  const user = users.find((candidate) => candidate.id === id);
  if (!user) {
    throw new NotFoundError(id);
  }
  return user;
}

for (const id of [1, 99]) {
  try {
    console.log(findUser(id).name);
  } catch (error) {
    if (error instanceof NotFoundError) {
      console.log("missing id:", error.id);
    } else {
      throw error;
    }
  }
}

// Re-throwing matters because a typo inside findUser would otherwise be caught
// here and reported as "user not found", hiding a real bug.
```
