# JavaScript — Destructuring & spread

## LESSON 31 — Destructuring, rest, spread

This is the syntax React code is written in. Learn it here and React will look familiar.

### Destructuring an object

Pull values out into variables named after the keys.

```js
const user = { name: "Sam", age: 34, city: "Geneva" };

const { name, age } = user;   // name -> "Sam", age -> 34
```

You can rename and set defaults:

```js
const { name: userName, country = "Italy" } = user;
// userName -> "Sam", country -> "Italy" (key missing, default used)
```

### Destructuring an array

Here **position** matters, not names.

```js
const colors = ["red", "green", "blue"];

const [first, second] = colors;   // "red", "green"
```

### Rest — collect what's left

```js
const [head, ...others] = colors;      // head -> "red", others -> ["green", "blue"]
const { name, ...rest } = user;        // rest -> { age: 34, city: "Geneva" }
```

### Spread — unpack into a new thing

```js
const copy = [...colors];                    // a real copy
const more = [...colors, "yellow"];          // add without modifying
const updated = { ...user, city: "Rome" };   // copy + overwrite one key
```

### Destructuring in parameters

```js
function greet({ name, age }) {
  return `${name} is ${age}`;
}

greet(user);
```

### Key notes

- **`const copy = original` does NOT copy** an array or an object. Both names point at the same thing, so changing one changes the other. `[...original]` and `{ ...original }` make a real copy.
- Same three dots, two opposite jobs: on the **left** of `=` it's **rest** (collect), on the **right** it's **spread** (unpack).
- Object destructuring matches by **key name**, array destructuring by **position**.
- In `{ ...user, city: "Rome" }` order matters: the last one wins.

### Example

````typescript
const exampleUser = { name: "Sam", age: 34, city: "Geneva" };

const { name, age } = exampleUser;
console.log(name, age);

const { name: userName, country = "Italy" } = exampleUser;
console.log(userName, country);

const exampleColors = ["red", "green", "blue"];
const [firstColor, secondColor] = exampleColors;
console.log(firstColor, secondColor);

const [head, ...others] = exampleColors;
console.log(head, others);

const { name: _skip, ...rest } = exampleUser;
console.log(rest);

// Spread makes a real copy — the original stays untouched
const exampleCopy = [...exampleColors, "yellow"];
console.log(exampleCopy);
console.log(exampleColors);

const exampleUpdated = { ...exampleUser, city: "Rome" };
console.log(exampleUpdated);

// Without spread: same object, not a copy
const exampleAlias = exampleUser;
exampleAlias.age = 99;
console.log(exampleUser.age);   // 99 — surprised?

function exampleGreet({ name, age }) {
  return `${name} is ${age}`;
}

console.log(exampleGreet(exampleUser));
````
