### LESSON 31 — Exercise

```typescript
const product = { name: "Laptop", price: 1200, stock: 5 };

const { name, price } = product;
console.log(name, price);

const { stock: available } = product;
console.log(available);

const discounted = { ...product, price: 999 };

console.log(discounted);
console.log(product);
```
