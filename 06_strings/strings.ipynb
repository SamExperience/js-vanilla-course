# JavaScript — Strings

## LESSON 22 — Reading and searching strings

A string is a sequence of characters, and every character has an index — exactly like an array.

```js
const city = "Geneva";

city.length;      // 6
city[0];          // "G"
city.at(-1);      // "a"   -> at() counts from the end when given a negative number
```

### Searching

| method | gives back | question it answers |
|---|---|---|
| `includes(text)` | `true` / `false` | is it in there at all? |
| `indexOf(text)` | index, or `-1` | where does it start? |
| `startsWith(text)` | `true` / `false` | does it begin with this? |
| `endsWith(text)` | `true` / `false` | does it end with this? |

```js
const email = "sam@example.com";

email.includes("@");        // true
email.indexOf("@");         // 3
email.endsWith(".com");     // true
```

### Cutting a piece out

`slice(start, end)` returns the characters from `start` up to **but not including** `end`.

```js
const code = "IT-2026-004";

code.slice(0, 2);     // "IT"
code.slice(3);        // "2026-004"   -> no end means "to the end"
code.slice(-3);       // "004"        -> negative counts from the end
```

### Template literals

Backticks let you drop values straight into text with `${}`. This is how you will build almost every message from here on.

```js
const name = "Mia";
const age = 28;

`${name} is ${age}`;        // "Mia is 28"
`Next year: ${age + 1}`;    // "Next year: 29"   -> any expression works inside ${}
```

### Key notes

- **Strings never change.** `city.toUpperCase()` does not modify `city`, it returns a new string. If you do not store the result, nothing happened.
- `indexOf` returns **-1** when the text is missing, not `undefined`. And `-1` is truthy, so `if (str.indexOf("a"))` is a bug — use `includes` when you want a yes or no answer.
- `slice(0, 2)` gives you **2** characters, not 3. The end index is excluded, the same rule as everywhere else in JavaScript.
- Template literals use **backticks**, not quotes. `"${name}"` prints the dollar sign and the braces literally.

### Example

````typescript
const exampleFile = "report-2026.pdf";

console.log(exampleFile.length);
console.log(exampleFile[0]);
console.log(exampleFile.at(-1));

console.log(exampleFile.includes("2026"));
console.log(exampleFile.indexOf("-"));
console.log(exampleFile.endsWith(".pdf"));

console.log(exampleFile.slice(0, 6));
console.log(exampleFile.slice(-3));

const exampleUser = "Mia";
console.log(`Welcome back, ${exampleUser}!`);
````

## LESSON 23 — Transforming strings

Every method here returns a **new** string. The original is untouched.

| method | what it does |
|---|---|
| `toUpperCase()` / `toLowerCase()` | changes the case |
| `trim()` | removes spaces at both ends |
| `replace(a, b)` | replaces the **first** `a` with `b` |
| `replaceAll(a, b)` | replaces **every** `a` with `b` |
| `padStart(n, char)` | pads at the front until the string is `n` long |
| `repeat(n)` | repeats the string `n` times |

```js
"  Sam  ".trim();                    // "Sam"
"hello world".toUpperCase();         // "HELLO WORLD"
"a-b-c".replace("-", "+");           // "a+b-c"     -> only the first
"a-b-c".replaceAll("-", "+");        // "a+b+c"
"7".padStart(2, "0");                // "07"
```

### split and join

`split` turns a string into an array. `join` turns an array back into a string. They are opposites, and together they are how most text gets processed.

```js
const line = "bread,milk,eggs";

const items = line.split(",");     // ["bread", "milk", "eggs"]
items.join(" | ");                 // "bread | milk | eggs"

"Sam".split("");                   // ["S", "a", "m"]   -> "" splits every character
```

### Chaining

Because each method returns a string, you can call the next one straight away.

```js
"  Hello World  ".trim().toLowerCase().replaceAll(" ", "-");   // "hello-world"
```

That is a URL slug, in one line.

### Key notes

- **`replace` only replaces the first match.** This surprises everyone once. Use `replaceAll` when you mean all of them.
- `trim()` removes spaces at the **ends**, never in the middle. `" a b ".trim()` is `"a b"`.
- `split("")` splits into single characters, `split(" ")` splits into words, and `split()` with no argument gives you a one-item array holding the whole string.
- Chaining reads left to right, and each step works on the result of the one before. If a chain misbehaves, print the intermediate steps one at a time.

### Example

````typescript
const exampleRaw = "  JavaScript Is Fun  ";

console.log(exampleRaw.trim());
console.log(exampleRaw.trim().toLowerCase());
console.log(exampleRaw.trim().toLowerCase().replaceAll(" ", "-"));

const exampleCsv = "red,green,blue";
const exampleColours = exampleCsv.split(",");
console.log(exampleColours);
console.log(exampleColours.join(" and "));

console.log("5".padStart(3, "0"));
console.log("ab".repeat(3));
````
