### LESSON 23 — Mini challenge

```typescript
function initials(fullName) {
  const parts = fullName.split(" ");
  const letters = parts.map((part) => part[0].toUpperCase());
  return letters.join(".") + ".";
}

console.log(initials("mia rossi"));
console.log(initials("samuele deriu"));

function censor(text, word) {
  return text.replaceAll(word, "*".repeat(word.length));
}

console.log(censor("bad news, bad day", "bad"));
```
