### LESSON 25 — Exercise, alternative version

Note the difference from the official solution: `pop()` removes the **last** item, while step 5 asks for the **first** one — that is what `shift()` is for.

```typescript
// Wrapped in a block so these names don't clash with the solution above.
{
  let shoppingList=["bread","milk","eggs"];
  console.log(shoppingList[0]);
  console.log(shoppingList.length);
  console.log(shoppingList);
  shoppingList.push("coffee");
  console.log(shoppingList);
  console.log(shoppingList.length);
  shoppingList.pop()
  console.log(shoppingList);
  console.log(shoppingList.length);
  console.log(shoppingList);
}
```

### LESSON 26 — Exercise

```typescript
const scores = [42, 7, 100, 8];

console.log(scores.slice().sort((a, b) => a - b));
console.log(scores);

console.log(scores.slice().sort((a, b) => b - a));

console.log(scores.slice().sort());
// Sorted as text: "100" comes before "42" because "1" comes before "4".

console.log(scores.slice(1, 3));
console.log(scores.at(-1));
```

### LESSON 26 — Mini challenge

```typescript
const queue = ["mia", "alex", "sam", "leo"];

// indexOf finds the position, splice removes 1 item starting there
queue.splice(queue.indexOf("alex"), 1);
console.log(queue);

// removing 0 items at leo's position inserts nina just before him
queue.splice(queue.indexOf("leo"), 0, "nina");
console.log(queue);

const numbered = [];
let position = 1;
for (const name of queue) {
  numbered.push(`${position}. ${name}`);
  position++;
}
console.log(numbered.join(", "));

// reverse a copy, so queue itself stays in order
console.log(queue.slice().reverse());
console.log(queue);
```
