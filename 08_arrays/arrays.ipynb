# JavaScript — Arrays

## LESSON 25 — Arrays

An **array** is an ordered list of values, written between square brackets.

```js
const fruits = ["Apple", "Banana", "Orange"];
```

### Accessing values

Each value has a position called **index**. Indexes start at **0**, not 1.

```js
fruits[0];   // "Apple"
fruits[2];   // "Orange"
fruits[9];   // undefined  (nothing there)
```

`.length` tells you how many items the array holds.

```js
fruits.length;              // 3
fruits[fruits.length - 1];  // "Orange"  -> the last item
```

### Adding and removing

| method | what it does |
|---|---|
| `push(value)` | adds at the **end** |
| `pop()` | removes the **last** one and returns it |
| `unshift(value)` | adds at the **beginning** |
| `shift()` | removes the **first** one and returns it |

### Searching

```js
fruits.includes("Banana");   // true  -> is it there?
fruits.indexOf("Banana");    // 1     -> at which index?
fruits.indexOf("Kiwi");      // -1    -> not found
```

### Key notes

- **The last index is `length - 1`.** With 3 items the indexes are 0, 1, 2 — never 3.
- `indexOf` returns **-1** when the value is missing, not `undefined` and not `false`.
- An array declared with `const` can still be **modified** (`push`, `pop`…). What `const` forbids is *reassigning* the variable to a different array.

### Example

````typescript
const exampleFruits = ["Apple", "Banana", "Orange"];

console.log(exampleFruits[0]);
console.log(exampleFruits.length);
console.log(exampleFruits[exampleFruits.length - 1]);

exampleFruits.push("Kiwi");
console.log(exampleFruits);

const exampleRemoved = exampleFruits.pop();
console.log(exampleRemoved);
console.log(exampleFruits);

console.log(exampleFruits.includes("Banana"));
console.log(exampleFruits.indexOf("Banana"));
console.log(exampleFruits.indexOf("Mango"));
````

## LESSON 26 — Reordering and slicing arrays

LESSON 25 covered adding and removing at the ends. These methods work anywhere in the array — and half of them change the original while the other half do not. That split is the thing to learn here.

| method | changes the original? | gives back |
|---|---|---|
| `slice(a, b)` | **no** | a copy of one section |
| `splice(i, n)` | **yes** | the removed items |
| `concat(other)` | no | a new joined array |
| `join(sep)` | no | a string |
| `reverse()` | **yes** | the same array, reversed |
| `sort()` | **yes** | the same array, sorted |
| `at(i)` | no | one item, negatives count from the end |

### slice — take a copy of a section

You already used `slice` on strings in LESSON 23. On arrays it is the same method with the same rules — it just hands you back an array instead of a string.

`slice(start, end)` takes two indexes, and **`start` is included while `end` is excluded**.

```
index:   0    1    2    3
        "a"  "b"  "c"  "d"
```

```js
const letters = ["a", "b", "c", "d"];

letters.slice(1, 3);   // ["b", "c"]             -> starts at 1, stops BEFORE 3
letters;               // ["a", "b", "c", "d"]   -> untouched
```

A quick way to check yourself: `slice(1, 3)` gives you `3 - 1 = 2` items.

Three shortcuts worth knowing:

```js
letters.slice(2);    // ["c", "d"]               -> from index 2 to the end
letters.slice(-2);   // ["c", "d"]               -> the last two
letters.slice();     // ["a", "b", "c", "d"]     -> a full copy
```

That last one matters for the rest of this lesson: `slice()` with no arguments is how you get a **copy** you can safely mess with.

### splice — cut and paste inside the array

`splice` looks like `slice` and behaves nothing like it. The trap is the second number.

**In `splice(start, howMany)` the second number is how many items to remove — not an end index.**

```js
const items = ["a", "b", "c", "d"];

items.splice(1, 2);   // ["b", "c"]   -> the removed items come back
items;                // ["a", "d"]   -> the original was cut
```

Pass `0` as `howMany` and nothing is removed — then anything you list after it gets **inserted** at that position:

```js
const gaps = ["a", "d"];

gaps.splice(1, 0, "b", "c");   // []                       -> nothing removed
gaps;                          // ["a", "b", "c", "d"]
```

Remove one and insert one in the same call, and you have replaced an item:

```js
const colors = ["red", "gren", "blue"];

colors.splice(1, 1, "green");   // ["gren"]
colors;                         // ["red", "green", "blue"]
```

Side by side, both starting from `["a", "b", "c", "d"]`:

| | `slice(1, 3)` | `splice(1, 3)` |
|---|---|---|
| the second number is | an end index, **excluded** | **how many** to remove |
| you get back | `["b", "c"]` — the copied section | `["b", "c", "d"]` — the removed items |
| the original becomes | `["a", "b", "c", "d"]` — untouched | `["a"]` — cut |

If you ever mix them up, print the original array right after the call. That one line tells you which of the two you just used.

### reverse — flip the order

`reverse` turns the array back to front. It changes the original and returns that same array — it does **not** hand you a new one.

```js
const letters = ["a", "b", "c"];

letters.reverse();   // ["c", "b", "a"]
letters;             // ["c", "b", "a"]   -> the original changed too
```

To keep the original in order, reverse a copy:

```js
const names = ["mia", "alex", "sam"];

const backwards = names.slice().reverse();   // ["sam", "alex", "mia"]
names;                                       // ["mia", "alex", "sam"]   -> still in order
```

### sort, and its trap

With no argument, `sort` compares items **as text**.

```js
[2, 10].sort();          // [10, 2]        <- wrong
[10, 9, 100].sort();     // [10, 100, 9]   <- wrong
```

Nothing is broken. The engine turned the numbers into text first, and `"10"` really does come before `"2"`, the same way "ap" comes before "b" in a dictionary.

To sort numbers you have to say how to compare them, by handing `sort` a function — the same "a function is a value you can pass around" idea from LESSON 21:

```js
[10, 9, 100].sort((a, b) => a - b);   // [9, 10, 100]
```

Here is the deal `sort` makes with that function. It hands you **two items at a time** and reads the number you give back:

| you return | what sort does |
|---|---|
| a negative number | puts `a` first |
| `0` | a tie, leaves them as they are |
| a positive number | puts `b` first |

So with `a = 2` and `b = 10`, `a - b` is `-8`, which is negative, so `2` goes first. Smaller first, all the way down the array — that is ascending order. Flip the subtraction and you flip the result:

```js
numbers.sort((a, b) => a - b);   // ascending
numbers.sort((a, b) => b - a);   // descending
```

For plain text you don't need any of this, because the default already sorts alphabetically:

```js
["pear", "apple", "fig"].sort();   // ["apple", "fig", "pear"]
```

Sorting by a field instead — a list of people by age, say — needs objects, and those come in LESSON 28.

### Keeping the original

`sort` and `reverse` change the array you call them on. When you need to keep it, copy it first and work on the copy:

```js
const scores = [42, 7, 100];

const ranked = scores.slice().sort((a, b) => a - b);   // [7, 42, 100]
scores;                                                // [42, 7, 100]   -> safe
```

You may see `[...scores]` written instead of `scores.slice()`. It is the same `...` you used for rest parameters in LESSON 20, working in the opposite direction — LESSON 31 covers it properly.

### concat, join and at

`concat` joins two arrays into a **new** one. Neither original is touched.

```js
["x", "y"].concat(["z"]);   // ["x", "y", "z"]
```

`join` is the one from LESSON 23, seen from the other side: it glues the items into a single string, using the separator you pass.

```js
["x", "y", "z"].join(" - ");   // "x - y - z"
```

`at` reads one item and accepts **negative** indexes, counting from the end. It is the readable version of the `length - 1` arithmetic from LESSON 25.

```js
const items = [1, 2, 3];

items.at(0);    // 1
items.at(-1);   // 3   -> same as items[items.length - 1]
```

### Key notes

- **`slice` copies, `splice` cuts.** One letter apart, opposite effects. Print the original right after the call to see which one you used.
- **The second number means two different things.** In `slice(1, 3)` it is the end index, and it is excluded. In `splice(1, 3)` it is how many items to remove.
- **`sort` and `reverse` change the original.** If you need to keep it, work on a copy: `items.slice().sort(...)`.
- **`sort()` with no argument sorts numbers wrongly**, silently and without an error. `[2, 10]` becomes `[10, 2]` because `"10"` comes before `"2"` as text.
- The function you hand to `sort` gives back a **number**, not `true` or `false`: negative to put `a` first, positive to put `b` first.

### Example

````typescript
const exampleLetters = ["a", "b", "c", "d"];

// slice copies a section and never touches the original
console.log(exampleLetters.slice(1, 3));
console.log(exampleLetters.slice(2));
console.log(exampleLetters.slice(-2));
console.log(exampleLetters.slice());
console.log(exampleLetters);

// splice cuts the original: work on a copy to keep the first array intact
const exampleCut = exampleLetters.slice();
console.log(exampleCut.splice(1, 2));
console.log(exampleCut);

// splice with 0 removes nothing and inserts instead
const exampleGaps = ["a", "d"];
exampleGaps.splice(1, 0, "b", "c");
console.log(exampleGaps);

// remove one, insert one: a replacement
const exampleColors = ["red", "gren", "blue"];
console.log(exampleColors.splice(1, 1, "green"));
console.log(exampleColors);

// reverse changes the original, so reverse a copy when you need both
const exampleNames = ["mia", "alex", "sam"];
console.log(exampleNames.slice().reverse());
console.log(exampleNames);
exampleNames.reverse();
console.log(exampleNames);

// sort compares as text unless you say how to compare
console.log([2, 10].sort());
console.log([10, 9, 100].sort());
console.log([10, 9, 100].sort((a, b) => a - b));
console.log([10, 9, 100].sort((a, b) => b - a));
console.log(["pear", "apple", "fig"].sort());

console.log(["x", "y"].concat(["z"]));
console.log(["x", "y", "z"].join(" - "));
console.log([1, 2, 3].at(-1));
````
