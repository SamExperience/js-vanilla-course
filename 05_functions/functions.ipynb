# JavaScript — Functions

## LESSON 19 — Functions

A function is a reusable block of code.

**Basic syntax**

```js
function functionName() {
    // code
}
```

Call the function:

```js
functionName();
```

### Parameters

Parameters are values received by the function.

```js
function greet(name) {
    console.log("Hello " + name);
}

greet("Sam");
```

`name` is the **parameter**. `"Sam"` is the **argument**.

### Return

`return` sends a value back to the caller.

```js
function add(a, b) {
    return a + b;
}

const result = add(2, 3);   // result -> 5
```

### Key notes

- **Parameter** → the variable defined by the function.
- **Argument** → the actual value passed to the function.
- `return` → gives a result back to the caller.
- `console.log()` → prints something to the output.

These two are **not** the same:

```js
console.log(value);   // prints it, gives nothing back
return value;         // hands it back to whoever called the function
```

### Example

Run this cell to see both patterns in action.

````typescript
function exampleGreet(name) {
  console.log("Hello " + name);
}

exampleGreet("Sam");

function exampleAdd(firstNumber, secondNumber) {
  return firstNumber + secondNumber;
}

const exampleSum = exampleAdd(10, 5);

console.log(exampleSum);
````

## LESSON 20 — Parameters in depth

### Default values

A parameter can carry a fallback, used only when the argument is missing.

```js
function greet(name, greeting = "Hello") {
  return `${greeting}, ${name}`;
}

greet("Mia");              // "Hello, Mia"
greet("Mia", "Welcome");   // "Welcome, Mia"
```

The default fires for `undefined` only. Pass `null` and you get `null`.

Put parameters with defaults **last**, otherwise callers have to pass `undefined` on purpose to skip them.

### Rest parameters

`...` in a parameter list collects every remaining argument into a real array.

```js
function total(...prices) {
  return prices.reduce((sum, price) => sum + price, 0);
}

total(10, 20, 30);     // 60
total();               // 0
```

It can follow named parameters, but only as the last one:

```js
function describe(label, ...values) {
  return `${label}: ${values.length} items`;
}
```

### Too many, too few

JavaScript never complains about the number of arguments. Extra ones are ignored; missing ones are `undefined`. That is why defaults matter — nothing else is checking for you.

### Key notes

- **A default only replaces `undefined`.** `greet("Mia", null)` returns `null, Mia`, because `null` is a value you chose to pass.
- **Rest must come last.** `function f(...values, last)` is a syntax error, and the message says so.
- A default can use an earlier parameter: `function f(a, b = a * 2)`. Reading right to left does not work — `b` does not exist yet when `a` is being set up.
- Rest gives you a **real array**, so `map`, `filter` and `reduce` all work on it immediately.

### Example

````typescript
function exampleGreet(name, greeting = "Hello") {
  return `${greeting}, ${name}`;
}

console.log(exampleGreet("Mia"));
console.log(exampleGreet("Mia", "Welcome"));
console.log(exampleGreet("Mia", null));

function exampleTotal(...prices) {
  return prices.reduce((sum, price) => sum + price, 0);
}

console.log(exampleTotal(10, 20, 30));
console.log(exampleTotal());

function exampleDescribe(label, ...values) {
  return `${label}: ${values.length} items`;
}

console.log(exampleDescribe("Cart", "pen", "book", "lamp"));
````

## LESSON 21 — Functions as values

A function in JavaScript is a value. You can store it in a variable, pass it to another function, and return it from one. Almost everything you write from here on relies on that.

### Arrow functions

A shorter way to write a function expression.

```js
function doubleOld(n) {
  return n * 2;
}

const double = (n) => n * 2;     // same thing
```

- One expression after `=>` is returned automatically.
- Add braces and you are writing a normal body, so you need `return` yourself.

```js
const double = (n) => n * 2;        // returns n * 2
const broken = (n) => { n * 2; };   // returns undefined
```

### Passing a function to a function

A function given as an argument is called a **callback**. The receiver decides when to call it.

```js
function repeat(times, action) {
  for (let i = 1; i <= times; i++) {
    action(i);
  }
}

repeat(3, (i) => console.log(`run ${i}`));
```

You have already used this shape without naming it: `addEventListener` and `map` both take a callback.

### Returning a function

A function that builds another function lets you fix part of the work in advance.

```js
function multiplier(factor) {
  return (n) => n * factor;
}

const triple = multiplier(3);
triple(5);                       // 15
```

`triple` remembers `factor` even though `multiplier` has finished. That memory is called a **closure**, and it is why the click counter in a later lesson can keep counting between clicks.

### Key notes

- **Pass the function, do not call it.** `repeat(3, action)` hands over the function; `repeat(3, action())` runs it now and hands over its result — usually `undefined`.
- **`(n) => { n * 2 }` returns `undefined`.** Braces mean a body, and a body needs `return`. This is the single most common arrow-function bug.
- A function stored in a `const` cannot be used before that line runs. A `function` declaration can. That is the practical difference between the two forms.
- A closure keeps the **variable**, not a copy of its value. If the variable changes later, the closure sees the change.

### Example

````typescript
const exampleDouble = (n) => n * 2;
const exampleBroken = (n) => { n * 2; };

console.log(exampleDouble(4));
console.log(exampleBroken(4));

function exampleRepeat(times, action) {
  for (let i = 1; i <= times; i++) {
    action(i);
  }
}

exampleRepeat(3, (i) => console.log(`run ${i}`));

function exampleMultiplier(factor) {
  return (n) => n * factor;
}

const exampleTriple = exampleMultiplier(3);
console.log(exampleTriple(5));
console.log(exampleMultiplier(10)(5));
````
