### LESSON 19 — Mini challenge, alternative version

Note: `return age >= 18 ? true : false` works, but the comparison is already a boolean — `return age >= 18` says the same thing.

```typescript
// Wrapped in a block so these names don't clash with the solution above.
{
  function isAdult(age){
      return age >= 18 ? true : false;
  }
  let adult=isAdult(25);
  console.log(adult);
}
```

### LESSON 20 — Exercise

```typescript
function makeTag(text, tag = "p") {
  return `<${tag}>${text}</${tag}>`;
}

console.log(makeTag("Hello"));
console.log(makeTag("Hello", "h1"));

function average(...numbers) {
  if (numbers.length === 0) return 0;
  return numbers.reduce((sum, n) => sum + n, 0) / numbers.length;
}

console.log(average(10, 20, 30));
console.log(average());
```

### LESSON 20 — Mini challenge

```typescript
function buildUrl(base, ...parts) {
  const cleanBase = base.replace(/\/+$/, "");
  const cleanParts = parts.map((part) => part.replaceAll("/", ""));
  return [cleanBase, ...cleanParts].join("/");
}

console.log(buildUrl("https://api.dev", "users", "7"));
console.log(buildUrl("https://api.dev/", "/users/", "7"));
console.log(buildUrl("https://api.dev"));
```

### LESSON 21 — Exercise

```typescript
const square = (n) => n * n;
console.log(square(4));

function applyTwice(value, fn) {
  return fn(fn(value));
}

console.log(applyTwice(3, square));

function greeterFor(greeting) {
  return (name) => `${greeting}, ${name}`;
}

const hello = greeterFor("Hello");
const welcome = greeterFor("Welcome");

console.log(hello("Mia"));
console.log(welcome("Alex"));
```

### LESSON 21 — Mini challenge

```typescript
function makeCounter() {
  let count = 0;
  return () => {
    count += 1;
    return count;
  };
}

const next = makeCounter();
console.log(next(), next(), next());

const other = makeCounter();
console.log(other());

// The count lives in the call to makeCounter that produced the function.
// The returned function keeps that call's variables alive, so each counter
// gets its own private count.
```
