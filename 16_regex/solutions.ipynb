### LESSON 39 — Mini challenge

```typescript
function isValidPostcode(text) {
  return /^\d{5}$/.test(text);
}

console.log(isValidPostcode("09100"), isValidPostcode("0910"), isValidPostcode("09100x"));

function maskCard(number) {
  return number.replace(/\d(?=.*\d{4})/g, "*");
}

console.log(maskCard("1234 5678 9012 3456"));

function countWords(text) {
  const trimmed = text.trim();
  if (trimmed === "") return 0;
  return trimmed.split(/\s+/).length;
}

console.log(countWords("one   two three"), countWords("   "), countWords(""));
```
