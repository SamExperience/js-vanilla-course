# JavaScript — Regular expressions

## LESSON 39 — Regular expressions

A regular expression is a small pattern language for describing text. You write it between slashes.

```js
const pattern = /cat/;

pattern.test("concatenate");     // true    <- is it in there?
"a cat here".match(/cat/);       // the match, with its index
"a cat here".replace(/cat/, "dog");
```

Everything you already do with `includes` and `replaceAll` still works and is easier to read. Reach for a regex when the thing you are looking for has a **shape** rather than an exact spelling.

### The tokens worth knowing

| token | matches |
|---|---|
| `\d` | any digit |
| `\w` | any letter, digit or underscore |
| `\s` | any space, tab or newline |
| `.` | any character at all |
| `[abc]` | one of a, b or c |
| `[^abc]` | anything except a, b or c |
| `+` | one or more of what came before |
| `*` | zero or more |
| `?` | zero or one — optional |
| `{2,4}` | between two and four |
| `^` / `$` | start / end of the text |

```js
/^\d{4}$/.test("2026");        // true   -> exactly four digits, nothing else
/\d+/.exec("abc 42 def")[0];   // "42"
```

### Flags

A flag goes after the closing slash: `g` for every match instead of the first, `i` to ignore case.

```js
"Cat cat".replace(/cat/gi, "dog");     // "dog dog"
"a1b2c3".match(/\d/g);                // ["1", "2", "3"]
```

### Groups

Round brackets capture a piece so you can read it back.

```js
const match = "2026-09-10".match(/(\d{4})-(\d{2})-(\d{2})/);

match[1];    // "2026"
match[2];    // "09"
```

### Key notes

- **Use the simplest tool that works.** `includes` beats a regex when you know the exact text you want.
- **`test` with a `/g` regex is stateful.** It remembers where it stopped in `lastIndex`, so calling it twice on the same string can return `true` then `false`. Do not reuse a `/g` regex for repeated `test` calls.
- `match` without `/g` gives you the groups; `match` with `/g` gives you every match and **no** groups. Two different return shapes from the same method.
- Inside a character class, most special characters lose their meaning: `[.]` is a literal dot.

### Example

````typescript
console.log(/cat/.test("concatenate"));
console.log("a cat here".replace(/cat/, "dog"));
console.log("Cat cat".replace(/cat/gi, "dog"));

console.log(/^\d{4}$/.test("2026"));
console.log(/^\d{4}$/.test("20268"));

console.log("a1b2c3".match(/\d/g));

const exampleMatch = "2026-09-10".match(/(\d{4})-(\d{2})-(\d{2})/);
console.log(exampleMatch[1], exampleMatch[2], exampleMatch[3]);

const exampleEmail = /^[\w.]+@[\w.]+\.\w{2,}$/;
console.log(exampleEmail.test("sam@example.com"));
console.log(exampleEmail.test("not-an-email"));
````
