### LESSON 32 — Exercise

```typescript
const order = {
  id: 7,
  customer: { name: "Alex" },
  discount: 0,
};

console.log(order.customer.name);
console.log(order.shipping?.city);

console.log(order.discount ?? 10);   // 0 — the real value survives
console.log(order.discount || 10);   // 10 — wrong, 0 is falsy

order.note ??= "none";
console.log(order.note);
```
