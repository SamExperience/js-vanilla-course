### LESSON 33 — Exercise

```typescript
class Person {
  constructor(name, age) {
    this.name = name;
    this.age = age;
  }

  greet() {
    return `Hi, I'm ${this.name}`;
  }

  get isAdult() {
    return this.age >= 18;
  }
}

const adult = new Person("Mia", 28);
const child = new Person("Leo", 9);

console.log(adult.greet(), adult.isAdult);
console.log(child.greet(), child.isAdult);
```

### LESSON 33 — Mini challenge

```typescript
class Playlist {
  constructor() {
    this.songs = [];
  }

  add(title) {
    this.songs.push(title);
    return this;            // returning the object is what makes chaining work
  }

  get count() {
    return this.songs.length;
  }

  static fromArray(titles) {
    const playlist = new Playlist();
    titles.forEach((title) => playlist.add(title));
    return playlist;
  }
}

const chained = new Playlist().add("One").add("Two").add("Three");
const built = Playlist.fromArray(["A", "B"]);

console.log(chained.count);
console.log(built.count);
```

### LESSON 34 — Exercise

```typescript
class Shape {
  constructor(name) {
    this.name = name;
  }

  area() {
    return 0;
  }
}

class Rectangle extends Shape {
  constructor(width, height) {
    super("Rectangle");
    this.width = width;
    this.height = height;
  }

  area() {
    return this.width * this.height;
  }
}

class Square extends Rectangle {
  constructor(side) {
    super(side, side);
    this.name = "Square";
  }
}

const rectangle = new Rectangle(4, 5);
const square = new Square(3);

console.log(rectangle.name, rectangle.area());
console.log(square.name, square.area());
console.log(square instanceof Shape);
```

### LESSON 34 — Mini challenge

```typescript
class Employee {
  constructor(name, salary) {
    this.name = name;
    this.salary = salary;
  }

  report() {
    return `${this.name} earns ${this.salary}`;
  }
}

class Manager extends Employee {
  constructor(name, salary, teamSize) {
    super(name, salary);
    this.teamSize = teamSize;
  }

  report() {
    return `${super.report()} and leads ${this.teamSize} people`;
  }
}

const staff = [new Employee("Alex", 3000), new Manager("Mia", 5000, 4)];

for (const person of staff) {
  console.log(person.report());
}
```
